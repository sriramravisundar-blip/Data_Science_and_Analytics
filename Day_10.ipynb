## 5. Data Cleaning and Preparation

Handle missing values and prepare data for analysis.

```python
# Create a copy for analysis
df_clean = df.copy()

# Handle missing values in Delay_Minutes
print(f"Missing values in Delay_Minutes before handling: {df_clean['Delay_Minutes'].isnull().sum()}")

# Fill missing Delay_Minutes with 0 (assuming no delay means cancelled/missing data)
df_clean['Delay_Minutes'].fillna(0, inplace=True)

print(f"Missing values in Delay_Minutes after handling: {df_clean['Delay_Minutes'].isnull().sum()}")

# Convert Flight_Date to datetime
df_clean['Flight_Date'] = pd.to_datetime(df_clean['Flight_Date'])

# Add derived columns for analysis
df_clean['Seat_Utilization'] = (df_clean['Passengers'] / df_clean['Seat_Capacity'] * 100).round(2)
df_clean['Total_Revenue'] = (df_clean['Passengers'] * df_clean['Average_Ticket_Price']).round(2)
df_clean['Month'] = df_clean['Flight_Date'].dt.month
df_clean['Month_Name'] = df_clean['Flight_Date'].dt.strftime('%B')

print("\nData Cleaning Complete!")
print(f"Dataset shape after cleaning: {df_clean.shape}")
```

```python
# Display cleaned data sample
print("Sample of cleaned data with new columns:")
print(df_clean[['Flight_ID', 'Passengers', 'Seat_Capacity', 'Seat_Utilization', 'Total_Revenue', 'Month_Name']].head(10))
```

## 6. Data Filtering and Selection

Filter data based on specific criteria to extract meaningful subsets.

```python
# Filter 1: Flights with delays > 50 minutes
delayed_flights = df_clean[df_clean['Delay_Minutes'] > 50]
print(f"Flights with delays > 50 minutes: {len(delayed_flights)} flights")
print(delayed_flights[['Flight_ID', 'Airline', 'Delay_Minutes', 'Weather']].head())

print("\n" + "="*80 + "\n")

# Filter 2: Premium class flights
premium_flights = df_clean[df_clean['Travel_Class'].isin(['Premium Economy', 'Business'])]
print(f"Premium class flights: {len(premium_flights)} flights")
print(f"Percentage of premium flights: {(len(premium_flights)/len(df_clean)*100):.2f}%")

print("\n" + "="*80 + "\n")

# Filter 3: High utilization flights (>90%)
high_util = df_clean[df_clean['Seat_Utilization'] > 90]
print(f"High utilization flights (>90%): {len(high_util)} flights")
print(f"Average seat utilization: {high_util['Seat_Utilization'].mean():.2f}%")
```

```python
# Filter 4: On-time flights by specific airline
air_india_ontime = df_clean[(df_clean['Airline'] == 'Air India') & (df_clean['Flight_Status'] == 'On Time')]
print(f"Air India On-Time flights: {len(air_india_ontime)} out of {len(df_clean[df_clean['Airline'] == 'Air India'])} total Air India flights")
print(f"On-time percentage: {(len(air_india_ontime)/len(df_clean[df_clean['Airline'] == 'Air India'])*100):.2f}%")

print("\n" + "="*80 + "\n")

# Filter 5: Select specific columns for a subset
selected_cols = df_clean[['Flight_ID', 'Airline', 'Origin', 'Destination', 'Average_Ticket_Price', 'Passenger_Satisfaction']]
high_satisfaction = selected_cols[df_clean['Passenger_Satisfaction'] >= 5]
print(f"Flights with maximum satisfaction (5/5): {len(high_satisfaction)} flights")
print(high_satisfaction.head(10))
```

## 7. Sorting Operations

Sort data by different criteria to identify patterns and extremes.

```python
# Sort 1: Highest revenue flights
print("Top 10 Revenue-Generating Flights:")
top_revenue = df_clean.nlargest(10, 'Total_Revenue')[['Flight_ID', 'Airline', 'Passengers', 'Average_Ticket_Price', 'Total_Revenue']]
print(top_revenue)

print("\n" + "="*80 + "\n")

# Sort 2: Most delayed flights
print("Top 10 Most Delayed Flights:")
most_delayed = df_clean.nlargest(10, 'Delay_Minutes')[['Flight_ID', 'Airline', 'Delay_Minutes', 'Weather', 'Flight_Status']]
print(most_delayed)

print("\n" + "="*80 + "\n")

# Sort 3: Highest seat utilization
print("Top 10 Highest Seat Utilization Flights:")
high_util_sort = df_clean.nlargest(10, 'Seat_Utilization')[['Flight_ID', 'Airline', 'Passengers', 'Seat_Capacity', 'Seat_Utilization']]
print(high_util_sort)
```

```python
# Sort 4: Lowest average ticket price
print("\nFlights with Lowest Average Ticket Price:")
lowest_price = df_clean.nsmallest(10, 'Average_Ticket_Price')[['Flight_ID', 'Airline', 'Average_Ticket_Price', 'Origin', 'Destination']]
print(lowest_price)

print("\n" + "="*80 + "\n")

# Sort 5: Multiple sorting
print("Flights sorted by Flight_Status then by Delay_Minutes:")
multi_sort = df_clean.sort_values(by=['Flight_Status', 'Delay_Minutes'], ascending=[False, False])
print(multi_sort[['Flight_ID', 'Airline', 'Flight_Status', 'Delay_Minutes']].head(15))
```

## 8. Grouping and Aggregation

Group data by various criteria and compute aggregate statistics.

```python
# Grouping 1: By Airline
print("Aggregation by Airline:")
airline_agg = df_clean.groupby('Airline').agg({
    'Flight_ID': 'count',
    'Passengers': 'sum',
    'Average_Ticket_Price': 'mean',
    'Delay_Minutes': 'mean',
    'Total_Revenue': 'sum',
    'Passenger_Satisfaction': 'mean'
}).round(2)

airline_agg.columns = ['Total_Flights', 'Total_Passengers', 'Avg_Ticket_Price', 'Avg_Delay_Minutes', 'Total_Revenue', 'Avg_Satisfaction']
airline_agg = airline_agg.sort_values('Total_Revenue', ascending=False)
print(airline_agg)
```

```python
# Grouping 2: By Flight Status
print("\nAggregation by Flight Status:")
status_agg = df_clean.groupby('Flight_Status').agg({
    'Flight_ID': 'count',
    'Delay_Minutes': ['mean', 'max'],
    'Passenger_Satisfaction': 'mean',
    'Passengers': 'mean'
}).round(2)
print(status_agg)
```

```python
# Grouping 3: By Travel Class
print("\nAggregation by Travel Class:")
class_agg = df_clean.groupby('Travel_Class').agg({
    'Flight_ID': 'count',
    'Average_Ticket_Price': ['mean', 'min', 'max'],
    'Passengers': 'mean',
    'Passenger_Satisfaction': 'mean'
}).round(2)
print(class_agg)
```

```python
# Grouping 4: By Weather Conditions
print("\nAggregation by Weather Conditions:")
weather_agg = df_clean.groupby('Weather').agg({
    'Flight_ID': 'count',
    'Delay_Minutes': 'mean',
    'Flight_Status': lambda x: (x == 'Delayed').sum(),
    'Passenger_Satisfaction': 'mean'
}).round(2)
weather_agg.columns = ['Total_Flights', 'Avg_Delay_Minutes', 'Delayed_Flights', 'Avg_Satisfaction']
print(weather_agg)
```

```python
# Grouping 5: By Booking Channel and Travel Class
print("\nAggregation by Booking Channel and Travel Class:")
multi_group = df_clean.groupby(['Booking_Channel', 'Travel_Class']).agg({
    'Flight_ID': 'count',
    'Average_Ticket_Price': 'mean',
    'Passengers': 'mean'
}).round(2)
multi_group.columns = ['Total_Bookings', 'Avg_Price', 'Avg_Passengers']
print(multi_group)
```

```python
# Grouping 6: By Month
print("\nAggregation by Month:")
month_agg = df_clean.groupby('Month_Name').agg({
    'Flight_ID': 'count',
    'Passengers': 'sum',
    'Total_Revenue': 'sum',
    'Delay_Minutes': 'mean',
    'Passenger_Satisfaction': 'mean'
}).round(2)
month_agg.columns = ['Total_Flights', 'Total_Passengers', 'Total_Revenue', 'Avg_Delay', 'Avg_Satisfaction']
print(month_agg)
```

## 9. Data Transformations

Apply transformations to create new insights and features.

```python
# Transformation 1: Create delay category
def categorize_delay(minutes):
    if minutes == 0:
        return 'No Delay'
    elif minutes <= 30:
        return 'Minor Delay'
    elif minutes <= 60:
        return 'Moderate Delay'
    else:
        return 'Severe Delay'

df_clean['Delay_Category'] = df_clean['Delay_Minutes'].apply(categorize_delay)

print("Delay Category Distribution:")
print(df_clean['Delay_Category'].value_counts())
print(f"\nPercentage Distribution:")
print((df_clean['Delay_Category'].value_counts() / len(df_clean) * 100).round(2))
```

```python
# Transformation 2: Create seat utilization category
def categorize_utilization(util):
    if util < 50:
        return 'Low'
    elif util < 75:
        return 'Medium'
    elif util < 95:
        return 'High'
    else:
        return 'Very High'

df_clean['Utilization_Category'] = df_clean['Seat_Utilization'].apply(categorize_utilization)

print("Seat Utilization Category Distribution:")
print(df_clean['Utilization_Category'].value_counts())
print(f"\nAverage Satisfaction by Utilization Category:")
print(df_clean.groupby('Utilization_Category')['Passenger_Satisfaction'].mean().round(2))
```

```python
# Transformation 3: Create price category
def categorize_price(price):
    if price < 4000:
        return 'Budget'
    elif price < 7000:
        return 'Standard'
    else:
        return 'Premium'

df_clean['Price_Category'] = df_clean['Average_Ticket_Price'].apply(categorize_price)

print("Price Category Distribution:")
print(df_clean['Price_Category'].value_counts())
print(f"\nAverage Ticket Price by Category:")
print(df_clean.groupby('Price_Category')['Average_Ticket_Price'].mean().round(2))
```

```python
# Transformation 4: Create route column and analyze
df_clean['Route'] = df_clean['Origin'] + ' → ' + df_clean['Destination']

print("Top 10 Most Frequent Routes:")
route_freq = df_clean['Route'].value_counts().head(10)
print(route_freq)

print("\nRoute Analysis (Top 5 by Frequency):")
top_routes = route_freq.head(5).index.tolist()
for route in top_routes:
    route_data = df_clean[df_clean['Route'] == route]
    print(f"\nRoute: {route}")
    print(f"  - Flights: {len(route_data)}")
    print(f"  - Avg Passengers: {route_data['Passengers'].mean():.0f}")
    print(f"  - Avg Delay: {route_data['Delay_Minutes'].mean():.1f} minutes")
    print(f"  - Avg Satisfaction: {route_data['Passenger_Satisfaction'].mean():.2f}")
```

```python
# Transformation 5: Normalize/Scale numeric columns
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()
numeric_cols = ['Passengers', 'Average_Ticket_Price', 'Delay_Minutes', 'Avg_Baggage_Kg']

df_clean_normalized = df_clean.copy()
df_clean_normalized[numeric_cols] = scaler.fit_transform(df_clean[numeric_cols])

print("Normalized Data Sample (0-1 scale):")
print(df_clean_normalized[['Flight_ID', 'Passengers', 'Average_Ticket_Price', 'Delay_Minutes']].head(10))
```

## 10. Advanced Analysis and Insights

Answer meaningful business questions about the flight operations.

```python
# Question 1: Relationship between weather and delays
print("Question 1: How does weather impact flight delays?\n")
weather_delay = df_clean.groupby('Weather').agg({
    'Delay_Minutes': ['mean', 'median', 'max'],
    'Flight_ID': 'count'
}).round(2)
weather_delay.columns = ['Avg_Delay', 'Median_Delay', 'Max_Delay', 'Total_Flights']
print(weather_delay)
print(f"\nWeather with highest average delay: {df_clean.groupby('Weather')['Delay_Minutes'].mean().idxmax()}")
print(f"Weather with lowest average delay: {df_clean.groupby('Weather')['Delay_Minutes'].mean().idxmin()}")
```

```python
# Question 2: Passenger satisfaction across different airlines
print("\nQuestion 2: How does passenger satisfaction vary by airline?\n")
satisfaction_airline = df_clean.groupby('Airline')['Passenger_Satisfaction'].agg(['mean', 'median', 'std', 'count']).round(2)
satisfaction_airline.columns = ['Avg_Satisfaction', 'Median_Satisfaction', 'Std_Dev', 'Total_Flights']
satisfaction_airline = satisfaction_airline.sort_values('Avg_Satisfaction', ascending=False)
print(satisfaction_airline)
print(f"\nBest airline by satisfaction: {satisfaction_airline.index[0]}")
print(f"Lowest airline by satisfaction: {satisfaction_airline.index[-1]}")
```

```python
# Question 3: Revenue analysis
print("\nQuestion 3: What is the revenue potential by travel class?\n")
revenue_class = df_clean.groupby('Travel_Class').agg({
    'Total_Revenue': ['sum', 'mean'],
    'Flight_ID': 'count',
    'Passengers': 'sum'
}).round(2)
revenue_class.columns = ['Total_Revenue', 'Avg_Revenue_Per_Flight', 'Total_Flights', 'Total_Passengers']
print(revenue_class)
print(f"\nTotal Revenue Generated: ₹{df_clean['Total_Revenue'].sum():,.2f}")
print(f"Average Revenue per Flight: ₹{df_clean['Total_Revenue'].mean():,.2f}")
```

```python
# Question 4: Booking channel effectiveness
print("\nQuestion 4: Which booking channel is most effective?\n")
booking_analysis = df_clean.groupby('Booking_Channel').agg({
    'Flight_ID': 'count',
    'Passengers': 'sum',
    'Average_Ticket_Price': 'mean',
    'Passenger_Satisfaction': 'mean'
}).round(2)
booking_analysis.columns = ['Total_Bookings', 'Total_Passengers', 'Avg_Ticket_Price', 'Avg_Satisfaction']
booking_analysis = booking_analysis.sort_values('Total_Bookings', ascending=False)
print(booking_analysis)
print(f"\nMost popular booking channel: {booking_analysis.index[0]}")
```

```python
# Question 5: Operational efficiency metrics
print("\nQuestion 5: What are the key operational efficiency metrics?\n")
total_flights = len(df_clean)
on_time_flights = len(df_clean[df_clean['Flight_Status'] == 'On Time'])
delayed_flights = len(df_clean[df_clean['Flight_Status'] == 'Delayed'])
cancelled_flights = len(df_clean[df_clean['Flight_Status'] == 'Cancelled'])

print(f"Total Flights: {total_flights}")
print(f"On-Time Flights: {on_time_flights} ({on_time_flights/total_flights*100:.2f}%)")
print(f"Delayed Flights: {delayed_flights} ({delayed_flights/total_flights*100:.2f}%)")
print(f"Cancelled Flights: {cancelled_flights} ({cancelled_flights/total_flights*100:.2f}%)")

print(f"\nAverage Seat Utilization: {df_clean['Seat_Utilization'].mean():.2f}%")
print(f"Average Delay (all flights): {df_clean['Delay_Minutes'].mean():.2f} minutes")
print(f"Average Passenger Satisfaction: {df_clean['Passenger_Satisfaction'].mean():.2f}/5")
print(f"Average Baggage per Passenger: {df_clean['Avg_Baggage_Kg'].mean():.2f} kg")
```

```python
# Question 6: Peak travel periods
print("\nQuestion 6: When do peak travel periods occur?\n")
monthly_summary = df_clean.groupby('Month_Name').agg({
    'Flight_ID': 'count',
    'Passengers': 'sum',
    'Total_Revenue': 'sum'
}).round(2)
monthly_summary.columns = ['Total_Flights', 'Total_Passengers', 'Total_Revenue']
print(monthly_summary)
print(f"\nPeak passenger month: {df_clean.groupby('Month_Name')['Passengers'].sum().idxmax()}")
print(f"Peak revenue month: {df_clean.groupby('Month_Name')['Total_Revenue'].sum().idxmax()}")
```

```python
# Question 7: Aircraft utilization
print("\nQuestion 7: Which aircraft type is most utilized?\n")
aircraft_analysis = df_clean.groupby('Aircraft').agg({
    'Flight_ID': 'count',
    'Passengers': 'sum',
    'Total_Revenue': 'sum',
    'Delay_Minutes': 'mean'
}).round(2)
aircraft_analysis.columns = ['Total_Flights', 'Total_Passengers', 'Total_Revenue', 'Avg_Delay']
aircraft_analysis = aircraft_analysis.sort_values('Total_Flights', ascending=False)
print(aircraft_analysis)
print(f"\nMost utilized aircraft: {aircraft_analysis.index[0]}")
```
