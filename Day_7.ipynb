## Missing Values and Observations


```python
missing_values = df.isna().sum()
display(missing_values.to_frame(name='Missing Count'))

print('Total missing values:', missing_values.sum())

print('Unique values per column:')
display(df.nunique().to_frame(name='Unique Count'))
```

### Observations
- The dataset has 25 rows and 7 columns.
- Features include `Student_ID`, `Name`, `Age`, `Course`, `Marks`, `Attendance_Percentage`, and `City`.
- Numeric columns are `Student_ID`, `Age`, `Marks`, and `Attendance_Percentage`.
- Categorical/text columns are `Name`, `Course`, and `City`.
- There are missing values in `Marks`, `Attendance_Percentage`, and `City`, while the other columns are complete.
- The data appears to capture student records, including course enrollment, grades, attendance, and city of residence.
