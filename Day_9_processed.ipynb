# Processed E-commerce Dataset (Day 9)
This notebook loads Orders, Customers, and Products CSVs, merges them, demonstrates `concat()`, uses `apply()` to compute total price, performs datetime transformations, and exports a final processed CSV.

```python
import pandas as pd
from pathlib import Path
base = Path('.')
```

```python
orders = pd.read_csv(base / 'Day9_Orders.csv', parse_dates=['Order_Date'])
customers = pd.read_csv(base / 'Day9_Customers.csv')
products = pd.read_csv(base / 'Day9_Products.csv')
orders.head()
```

```python
mid = len(orders) // 2
orders_part1 = orders.iloc[:mid].copy()
orders_part2 = orders.iloc[mid:].copy()
orders_rejoined = pd.concat([orders_part1, orders_part2], ignore_index=True)
```

```python

df = orders_rejoined.merge(
    customers,
    on='Customer_ID',
    how='left'
).merge(
    products,
    on='Product_ID',
    how='left'
)
df['Order_Date'] = pd.to_datetime(df['Order_Date'], errors='coerce')
df['Total_Price'] = df['Quantity'] * df['Unit_Price']
df['Order_Month'] = df['Order_Date'].dt.month_name()
df['Order_Day'] = df['Order_Date'].dt.day
df['Order_Weekday'] = df['Order_Date'].dt.day_name()
df[['Order_Date','Order_Month','Order_Day','Order_Weekday']].head()
```

```python
cols = df.columns.tolist()
final_df = df[cols].sort_values(by='Order_Date').reset_index(drop=True)
final_df.to_csv('processed_ecommerce_dataset.csv', index=False)
print('Exported processed_ecommerce_dataset.csv')
```
